## Termin 01 | Eingabe, Regel, Ausgabe

Ein Versuch ist ein Wurf. Wir benutzen zuerst eine **feste** vorbereitete
Liste statt eines Zufallsgenerators. Die Modellregel lautet: Eine Sechs ist
ein Treffer. Zählen Sie die Sechsen zuerst selbst.

`=` speichert einen Wert unter einem Namen. Eckige Klammern bilden eine
Liste. `count(6)` zählt die Sechsen, `len` alle Einträge. `print` zeigt an.

```python
wuerfe = [2, 6, 1, 5, 6, 3, 4, 2, 6, 1]
treffer = wuerfe.count(6)
anzahl = len(wuerfe)
print(treffer, anzahl, treffer / anzahl)
```

Es erscheinen `3 10 0.3`: drei Treffer aus zehn Versuchen ergeben 30 Prozent.
Python schreibt Dezimalzahlen mit Punkt. Der Nenner enthält auch
Nichttreffer. Diese Liste prüft nur die Rechnung, nicht die Fairness eines
echten Würfels. Ein fairer Würfel hat Sechserwahrscheinlichkeit 1/6;
die beobachtete Quote einer kurzen Liste muss nicht genau 1/6 sein.

**Verständnisstopp:** Was würde passieren, wenn wir Treffer durch Treffer
teilen? Bei mindestens einem Treffer käme immer 1 heraus.

## Termin 02 | Den Zähler selbst schreiben

**Rückblick:** Ein Versuch ist ein Wurf, eine Liste enthält viele Versuche.
**Lernziele:** Listen durchlaufen, Funktionen mit Rückgabe verwenden und
die Ergebnisse an kleinen Testdaten prüfen. Vorwissen: Termin 01.

`for wurf in wuerfe` nimmt nacheinander jeden Eintrag. `if` prüft eine
Bedingung. `==` vergleicht zwei Werte; `=` weist einen Wert zu. Der
eingerückte Block gehört zur Schleife bzw. zur Bedingung.
**Vorhersage:** Wie entwickelt sich der Zähler bei `[2, 6, 1, 6]`?

```python
wuerfe = [2, 6, 1, 6]
treffer = 0
for wurf in wuerfe:
    vorher = treffer
    if wurf == 6:
        treffer = treffer + 1
    print("Wurf:", wurf, "vorher:", vorher, "nachher:", treffer)
print("Endstand:", treffer)
```

Die Zählerstände sind 0, 1, 1, 2. Der Startwert steht vor der Schleife,
sonst würden wir den bisherigen Stand immer wieder vergessen.
In dieser Demonstration zeigt `print` innerhalb der Schleife jeden Schritt;
die letzte Ausgabe steht außerhalb und erscheint nur einmal.

## Eine Funktion ist ein wiederverwendbarer Rechenweg
`def` definiert die Funktion. Die Liste kommt beim Aufruf hinein.
`return` gibt die berechnete Zahl zurück. `+= 1` ist die Kurzform von
`treffer = treffer + 1`.

```python
def zaehle_sechsen(wuerfe):
    treffer = 0
    for wurf in wuerfe:
        if wurf == 6:
            treffer += 1
    return treffer


anzahl_sechsen = zaehle_sechsen([6, 2, 6, 3, 6])
print("Rueckgabewert:", anzahl_sechsen)
print("Quote:", anzahl_sechsen / 5)
```

Die Funktion liefert 3 zurück; die anschließende Rechnung liefert 0.6.
**Unterschied:** `return` macht eine Zahl weiterverwendbar, `print` zeigt
etwas an. Ohne `return` liefert eine Funktion `None`, nicht die angezeigte
Zahl. Legen Sie `return` nicht in die Schleife: Sonst endet die Funktion
bereits vor den späteren Würfen.

## Kleine Tests
`assert` prüft eine Behauptung. Ist sie falsch, stoppt Python mit
`AssertionError`. Sind die folgenden Aussagen richtig, sehen wir die
Abschlussmeldung. Eine leere Liste hat null Treffer, aber keine definierte
Trefferquote: Durch null darf nicht geteilt werden.

```python
assert zaehle_sechsen([1, 2, 3]) == 0
assert zaehle_sechsen([6, 6]) == 2
assert zaehle_sechsen([]) == 0
print("Alle drei Tests bestanden")
```

**Brücke zum Mitteln:** Zählen addiert je Treffer 1; Summieren addiert

bedeutet Summe geteilt durch Anzahl, hier 4 / 2 = 2. In der Schleife

Fügen Sie eigene Codezellen ein. CR02.1/2 sind alternative Bearbeitungswege,

**Transfer, freiwillig:** Zählen Sie Werte mindestens gleich einer Grenze.
Erklären Sie an einem Randwert, welcher Vergleich passt.

**Drei Dinge bleiben:** Schleifen wiederholen Regeln; `return` gibt ein
Ergebnis zurück; kleine feste Daten helfen bei der Fehlersuche.
**Abschlussfrage:** Warum muss der Zähler vor der Schleife starten?
**Nächster Termin:** Die Auswertung bleibt gleich, die Eingaben werden gezogen.

```python
# A02.1
def zaehle_treffer(werte, ziel):
    treffer = 0

    for wert in werte:
        if wert == ziel:
            treffer = treffer + 1

    return treffer

print(zaehle_treffer([2, 6, 2, 1], 2))
print(zaehle_treffer([1, 3, 5], 2))
```

```python
# A02.2
def mittelwert(werte):
    summe = 0

    for wert in werte:
        summe = summe + wert

    return summe / len(werte)

print(mittelwert([2, 4, 6]))
print(mittelwert([5]))
print(mittelwert([-2, -4, -6]))
```
